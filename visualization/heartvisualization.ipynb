```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
os.chdir(r"C:\Users\lkhai\Downloads\DSBDA\visualization")
```

```python
heart = pd.read_csv("heart.csv", encoding='latin1')
```

```python
print("\nFIRST 5 ROWS OF DATASET")
print(heart.head())
```

```python
print(heart.info())
```

```python
print(heart.isnull().sum())
```

```python
numeric_heart = heart.select_dtypes(include='number')

print("\nNUMERIC COLUMNS")
print(numeric_heart.columns)
```

```python
plt.figure(figsize=(10,5))

sns.heatmap(
    heart.isnull(),
    cbar=False,
    cmap='viridis'
)

plt.title("Missing Values Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    numeric_heart.iloc[:,0],
    kde=True
)

plt.title("Distribution Graph")
plt.xlabel(numeric_heart.columns[0])

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x=numeric_heart.iloc[:,0]
)

plt.title("Outlier Detection Boxplot")
plt.xlabel(numeric_heart.columns[0])

plt.show()
```

```python
plt.figure(figsize=(12,8))

sns.heatmap(
    numeric_heart.corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")

plt.show()
```

```python

if len(numeric_heart.columns) >= 2:

    plt.figure(figsize=(8,5))

    sns.scatterplot(
        x=numeric_heart.iloc[:,0],
        y=numeric_heart.iloc[:,1]
    )

    plt.title("Scatter Plot")

    plt.xlabel(numeric_heart.columns[0])
    plt.ylabel(numeric_heart.columns[1])

    plt.show()
```

```python

numeric_heart.mean().plot(
    kind='bar',
    figsize=(10,5)
)

plt.title("Average Values of Numeric Columns")
plt.ylabel("Average")

plt.show()
```
