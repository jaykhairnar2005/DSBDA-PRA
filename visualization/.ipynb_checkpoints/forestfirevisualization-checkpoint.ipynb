```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
forest = pd.read_csv("forestfires.csv", encoding='latin1')
```

```python
print(forest.head())
```

```python
print(forest.info())
```

```python
print(forest.isnull().sum())
```

```python
numeric_forest = forest.select_dtypes(include='number')

print("\nNUMERIC COLUMNS")
print(numeric_forest.columns)
```

```python
plt.figure(figsize=(10,5))

sns.heatmap(
    forest.isnull(),
    cbar=False,
    cmap='viridis'
)

plt.title("Missing Values Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    numeric_forest.iloc[:,0],
    kde=True
)

plt.title("Distribution Graph")
plt.xlabel(numeric_forest.columns[0])

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x=numeric_forest.iloc[:,0]
)

plt.title("Outlier Detection Boxplot")
plt.xlabel(numeric_forest.columns[0])

plt.show()
```

```python
plt.figure(figsize=(12,8))

sns.heatmap(
    numeric_forest.corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
if len(numeric_forest.columns) >= 2:

    plt.figure(figsize=(8,5))

    sns.scatterplot(
        x=numeric_forest.iloc[:,0],
        y=numeric_forest.iloc[:,1]
    )

    plt.title("Scatter Plot")

    plt.xlabel(numeric_forest.columns[0])
    plt.ylabel(numeric_forest.columns[1])

    plt.show()
```

```python
if 'month' in forest.columns:

    plt.figure(figsize=(10,5))

    sns.countplot(
        x='month',
        data=forest
    )

    plt.title("Monthly Forest Fire Occurrences")

    plt.show()
```

```python
if 'wind' in forest.columns:

    plt.figure(figsize=(8,5))

    sns.boxplot(
        x=forest['wind']
    )

    plt.title("Wind Distribution")

    plt.show()
```

```python
numeric_forest.mean().plot(
    kind='bar',
    figsize=(12,5)
)

plt.title("Average Values of Numeric Columns")
plt.ylabel("Average")

plt.show()
```

```python
if 'temp' in forest.columns and 'area' in forest.columns:

    plt.figure(figsize=(8,5))

    sns.scatterplot(
        x='temp',
        y='area',
        data=forest
    )

    plt.title("Temperature vs Area Burned")

    plt.show()
```
