```python
page = requests.get(url, headers=headers)

print(page.status_code)
```

```python
soup = BeautifulSoup(page.content, "lxml")
```

```python
customer_name = []
ratings = []
review_title = []
reviews = []
```

```python
review_data = soup.find_all("div", {"data-hook": "review"})
```

```python
for item in review_data:
    
    # Customer Name
    name = item.find("span", class_="a-profile-name")
    customer_name.append(name.text.strip() if name else "N/A")
    
    # Rating
    rating = item.find("i", {"data-hook": "review-star-rating"})
    ratings.append(rating.text.strip() if rating else "N/A")
    
    # Review Title
    title = item.find("a", {"data-hook": "review-title"})
    review_title.append(title.text.strip() if title else "N/A")
    
    # Review Text
    review = item.find("span", {"data-hook": "review-body"})
    reviews.append(review.text.strip() if review else "N/A")
```

```python
data = pd.DataFrame({
    "Customer Name": customer_name,
    "Rating": ratings,
    "Review Title": review_title,
    "Review": reviews
})
```
