```python
import pandas as pd

df = pd.read_csv("dataset_Facebook.csv", sep=';')

print(df.columns)
```

```python
df.head()
```

```python
print("\n===== Data Subsets =====")

# Selecting specific columns
subset_columns = df[['Type', 'Category', 'like']]

print("\nColumn Subset:")
print(subset_columns.head())
```

```python
# Selecting rows where Type = Photo
subset_rows = df[df['Type'] == 'Photo']

print("\nRow Subset:")
print(subset_rows.head())
```

```python
print("\n===== Merge Data =====")

# Create another dataframe
df2 = pd.DataFrame({
    'Type': ['Photo', 'Status', 'Video'],
    'Description': ['Image Post', 'Text Post', 'Video Post']
})

# Merge both datasets
merged_data = pd.merge(df, df2, on='Type')

print(merged_data.head())
```

```python
print("\n===== Sort Data =====")

# Sort according to likes
sorted_data = df.sort_values(by='like', ascending=False)

print(sorted_data[['Type', 'like']].head())
```

```python
print("\n===== Transposed Data =====")

transposed_data = df.head().T

print(transposed_data)
```

```python
print("\n===== Shape of Dataset =====")

print(df.shape)
```

```python
print("\n===== Reshape Data using Melt =====")

melted_data = pd.melt(
    df,
    id_vars=['Type'],
    value_vars=['like', 'comment', 'share']
)

print(melted_data.head())


print("\n===== Pivot Table =====")

pivot_data = df.pivot_table(
    values='like',
    index='Type',
    columns='Category'
)

print(pivot_data)
```
