```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import seaborn as sns
```

```python
air = pd.read_csv("stations.csv")

heart = pd.read_csv("heart.csv")

forest = pd.read_csv("forestfires.csv")
```

```python
plt.figure(figsize=(6,4))

sns.countplot(x='AHD', data=heart)

plt.title("Heart Disease Count")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(heart['Age'], bins=10)

plt.title("Age Distribution")

plt.xlabel("Age")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x='AHD', y='Chol', data=heart)

plt.title("Cholesterol vs Heart Disease")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(x='temp', y='area', data=forest)

plt.title("Temperature vs Burned Area")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(forest['area'])

plt.title("Forest Fire Area Distribution")

plt.xlabel("Area")

plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.heatmap(forest.corr(numeric_only=True), cmap='YlGnBu')

plt.title("Forest Fire Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(air['elevation_m'])

plt.title("Elevation Distribution")

plt.xlabel("Elevation")

plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    x='coastal_proximity',
    y='valley_factor',
    data=air
)

plt.title("Coastal Proximity vs Valley Factor")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.heatmap(air.corr(numeric_only=True), cmap='coolwarm')

plt.title("Air Dataset Correlation Heatmap")

plt.show()
```
