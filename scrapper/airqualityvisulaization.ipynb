```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd

air = pd.read_csv("airquality_data.csv", encoding='latin1')


print(air.head())


```

```python
print(air.isnull().sum())
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(air.isnull(), cbar=False, cmap='viridis')
plt.title("Missing Values in Air Quality Dataset")
plt.show()
```

```python
# Select only numeric columns
numeric_air = air.select_dtypes(include='number')

# Show numeric columns
print(numeric_air.columns)
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    numeric_air.iloc[:,0],
    kde=True
)

plt.title("Distribution Graph")
plt.xlabel(numeric_air.columns[0])

plt.show()
```

```python
# Select numeric columns only
numeric_air = air.select_dtypes(include='number')

# Show numeric columns
print(numeric_air.columns)
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=numeric_air.iloc[:,0])

plt.title("Outlier Detection")
plt.xlabel(numeric_air.columns[0])

plt.show()
```

```python
plt.figure(figsize=(12,8))

sns.heatmap(numeric_air.corr(),
            annot=True,
            cmap='coolwarm')

plt.title("Correlation Heatmap")

plt.show()
```

```python
if len(numeric_air.columns) >= 2:

    plt.figure(figsize=(8,5))

    sns.scatterplot(
        x=numeric_air.iloc[:,0],
        y=numeric_air.iloc[:,1]
    )

    plt.title("Scatter Plot")

    plt.xlabel(numeric_air.columns[0])
    plt.ylabel(numeric_air.columns[1])

    plt.show()
```

```python
numeric_air.mean().plot(
    kind='bar',
    figsize=(10,5)
)

plt.title("Average Values of Numeric Columns")
plt.ylabel("Average")

plt.show()
```
